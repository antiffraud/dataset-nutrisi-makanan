```python
import json
import csv
from deep_translator import GoogleTranslator
from collections import Counter
```

```python
# 1. Baca file JSON asli
with open('fndds_nutrient_data3.json', 'r', encoding='utf-8') as file:
    data = json.load(file)
```

```python
# 2. Analisis data SEBELUM translate

# Total makanan
total_makanan = len(data)

# Total entri nutrien
total_nutrisi = sum(len(item['nutrients']) for item in data)

# Rata-rata nutrisi per makanan
rata_rata_nutrisi = total_nutrisi / total_makanan if total_makanan > 0 else 0

# Hitung frekuensi kemunculan nama nutrisi
nutrisi_counter = Counter()
for item in data:
    for nutrient in item['nutrients']:
        nutrisi_counter[nutrient['name']] += 1

top_5_nutrisi = nutrisi_counter.most_common(5)
```

```python
# 3. Tampilkan hasil analisis
print("\n📊 Statistik Nutrisi (sebelum diterjemahkan):")
print(f"Total makanan: {total_makanan}")
print(f"Total entri nutrisi: {total_nutrisi}")
print(f"Rata-rata nutrisi per makanan: {rata_rata_nutrisi:.2f}")
print("\n🥇 Top 5 nutrisi yang paling sering muncul:")
for i, (nutrient, count) in enumerate(top_5_nutrisi, 1):
    print(f"{i}. {nutrient} - {count} kali")
```

```python
# Tentukan jumlah data yang ingin diterjemahkan
jumlah_data_terjemahan = 700  # Misalnya, ambil 100 entri pertama

# 2. Ambil sejumlah entri food_name beserta nutrisi yang ada di dalamnya
data_terjemahan = data[100:jumlah_data_terjemahan]
```

```python
# 2. Inisialisasi translator
translator = GoogleTranslator(source='auto', target='id')
```

```python
# 4. Terjemahkan food_name dan nutrients
for item in data_terjemahan:
    # Translate food_name
    item['food_name'] = translator.translate(item['food_name'])

    # Translate setiap nama nutrisi di dalam nutrients
    for nutrient in item['nutrients']:
        nutrient['name'] = translator.translate(nutrient['name'])
```

```python
# 5. Simpan hasil terjemahan ke file JSON baru
with open('terjemah.json', 'w', encoding='utf-8') as outfile:
    json.dump(data_terjemahan, outfile, indent=2, ensure_ascii=False)

print("\n✅ Terjemahan otomatis selesai dan disimpan sebagai 'terjemah.json'")
```

```python
# 1. Baca data terjemahan
with open('terjemah.json', 'r', encoding='utf-8') as file:
    data = json.load(file)
```

```python
# 2. Buat kalimat deskriptif
deskripsi_data = []
for item in data:
    nama_makanan = item['food_name']
    deskripsi_nutrisi = []

    for nutrient in item['nutrients']:
        nama_nutrisi = nutrient['name'].split(',')[0].strip()  # ambil sebelum koma
        jumlah = nutrient['amount']
        satuan = nutrient['unit']
        deskripsi_nutrisi.append(f"{nama_nutrisi} {jumlah} {satuan}")

    # Tambahkan "dan" sebelum nutrisi terakhir
    if len(deskripsi_nutrisi) > 1:
        kalimat_nutrisi = ", ".join(deskripsi_nutrisi[:-1]) + f", dan {deskripsi_nutrisi[-1]}"
    elif deskripsi_nutrisi:
        kalimat_nutrisi = deskripsi_nutrisi[0]
    else:
        kalimat_nutrisi = "tidak diketahui"

    kalimat = f"{nama_makanan} memiliki kandungan nutrisi seperti: {kalimat_nutrisi}."
    deskripsi_data.append({
        "makanan": nama_makanan,
        "deskripsi_nutrisi": kalimat
    })
```

```python
# 3. Simpan ke CSV
with open('train.csv', 'w', newline='', encoding='utf-8') as csvfile:
    fieldnames = ['makanan', 'deskripsi_nutrisi']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    for row in deskripsi_data:
        writer.writerow(row)

print("\n📁 Kalimat deskripsi berhasil disimpan di 'train.csv'")
```
